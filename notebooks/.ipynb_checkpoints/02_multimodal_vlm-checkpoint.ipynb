# Stage 02: Multimodal VLM Prompt Engineering & Stochasticity

## Overview
This notebook evaluates local Vision-Language Models (VLMs) on biomedical microscopy images.
We compare an unconstrained naive prompt against a structured, descriptive prompt with strict JSON output and uncertainty handling.


```python
import sys
from pathlib import Path
import json

CURRENT_DIR = Path.cwd()
PROJECT_ROOT = CURRENT_DIR.parent if CURRENT_DIR.name.lower() == "notebooks" else CURRENT_DIR
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.vlm import (
    run_naive_vlm,
    run_structured_vlm,
    evaluate_vlm_stochasticity,
    NAIVE_PROMPT,
    STRUCTURED_VLM_PROMPT
)
```

## 1. Naive Prompting (Baseline)
In this test, the VLM is queried with an open-ended prompt without domain constraints.


```python
data_dir = PROJECT_ROOT / "data" / "nuclei_dataset"
if not data_dir.exists():
    data_dir = PROJECT_ROOT / "nuclei_dataset"

sample_img = data_dir / "train" / "images" / "train_000.png"
naive_res = run_naive_vlm(sample_img, temperature=0.7)
print("Naive Prompt Output:")
print(naive_res.get("raw_response", ""))
```

## 2. Structured Descriptive Prompting
Here, the model is anchored as purely descriptive (prohibiting ungrounded diagnosis) and forced into a strict JSON schema permitting 'uncertain'.


```python
structured_res = run_structured_vlm(sample_img, temperature=0.0)
print("Structured VLM JSON Output:")
print(json.dumps(structured_res.get("json", {}), indent=2))
```

## 3. Stochasticity vs Determinism Analysis
We evaluate repeated runs at temperature=0.7 (showing output variability) vs temperature=0.0 (showing strict determinism).


```python
stoch_runs = evaluate_vlm_stochasticity(sample_img, prompt=NAIVE_PROMPT, n_runs=3, temperature=0.7)
for r in stoch_runs:
    print(f"--- Run {r['run_index']} (Temp 0.7) ---")
    print(r['response'][:200], "...\n")

det_runs = evaluate_vlm_stochasticity(sample_img, prompt=STRUCTURED_VLM_PROMPT, n_runs=2, temperature=0.0)
for r in det_runs:
    print(f"--- Run {r['run_index']} (Temp 0.0) ---")
    print(r['response'][:200], "...\n")
```
